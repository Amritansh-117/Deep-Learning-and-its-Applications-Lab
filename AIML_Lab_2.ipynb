```python
#Amritansh Singh 23070521013
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Input: Hours Studied
x = np.array([1, 2, 3, 4, 5])

# Output: Marks Obtained
y = np.array([20, 40, 50, 65, 80])

print("Hours Studied:", x)
print("Marks Obtained:", y)

# Reshape input for scikit-learn
X = x.reshape(-1, 1)

# Create and train the model
model = LinearRegression()
model.fit(X, y)

print("\nModel Trained Successfully")
print("Slope (Coefficient):", model.coef_[0])
print("Intercept:", model.intercept_)

# Predict values
y_pred = model.predict(X)

print("\nActual Marks:", y)
print("Predicted Marks:", np.round(y_pred, 2))

# Evaluation Metrics
mae = mean_absolute_error(y, y_pred)
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)

print("\nPerformance Metrics")
print("Mean Absolute Error (MAE):", mae)
print("Mean Squared Error (MSE):", mse)
print("Root Mean Squared Error (RMSE):", rmse)

# Comparison Table
print("\nActual vs Predicted")
for actual, predicted in zip(y, y_pred):
    print(f"Actual: {actual:>3}   Predicted: {predicted:.2f}")

# Plot Regression Line
plt.figure(figsize=(7,5))
plt.scatter(X, y, color='blue', label='Actual Data')
plt.plot(X, y_pred, color='red', linewidth=2, label='Regression Line')
plt.xlabel("Hours Studied")
plt.ylabel("Marks Obtained")
plt.title("Simple Linear Regression")
plt.legend()
plt.grid(True)
plt.show()

# Predict Marks for a New Student
hours = np.array([[6]])
predicted_marks = model.predict(hours)

print(f"\nPredicted Marks for studying {hours[0][0]} hours: {predicted_marks[0]:.2f}")
```
